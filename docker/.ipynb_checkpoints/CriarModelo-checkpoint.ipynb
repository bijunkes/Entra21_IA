```python
import pandas as pd
import pymysql

connection = pymysql.connect(
    host='localhost',       
    port=3307,              # <-- Mudamos de 3306 para 3307
    user='ml_user',         
    password='ml_user',     # <-- Senha correspondente
    database='ml_db'        
)
```

```python
df = pd.read_csv('credit_data.csv')
```

```python
df = pd.drop('clienteid', axis=1, inplace=True)
```

```python
!pip install sqlalchemy
```

```python
from sqlalchemy import create_engine

engine = create_engine('mysql+pymysql://', creator=lambda: connection)
```

```python
df['age'].fillna(df['age'].median(), inplace=True)
```

```python
df.to_sql(name='emprestimos', con=engine, if_exists='append', index=False, chunksize=500)
```

```python
mediana = df['age].median()
df.loc[df['age'] < 0, 'age'] = mediana
```

```python
X = df.drop(['default'], axis=1).values
y = df['default'].values
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score
from sklearn.model_selection import train_test_split
```

```python
scaler = StandardScaler()
X = scaler.fit_transform(X)
```

```python
modelo = KNeighborsClassifier(n_neighbors=5)
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_forest=42)
```

```python
modelo.fit(X_train, y_train)
```

```python
predicao = modelo.predict(X_test)
accuracy_score(predicao, y_test)
```

```python
import joblib
```

```python
joblib.dump(modelo, 'modelo.joblib')
```
